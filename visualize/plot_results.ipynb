```python
import matplotlib.pyplot as plt
import numpy as np

# 数据
datasets = ['Changjiang', 'Huangpujiang', 'Sediment', 'HoneyNeg', 'HoneyPos', 'Synthetic']
methods = ['w/o cor', 'ours', 'ours v2', 'ours v3']
top1_values = [
    [0.4048, 0.3077, 0.4737, 0.3704, 0.4118, 0.3986],
    [0.4286, 0.3077, 0.4737, 0.3704, 0.4118, 0.3986],
    [0.4762, 0.2692, 0.5263, 0.4444, 0.4118, 0.4324],
    [0.4762, 0.2692, 0.5263, 0.4444, 0.4118, 0.4392]
]
top5_values = [
    [0.6429, 0.5, 0.5789, 0.4815, 0.5, 0.5608],
    [0.6667, 0.5, 0.5789, 0.5185, 0.5, 0.5676],
    [0.6905, 0.5385, 0.5789, 0.5185, 0.5294, 0.5743],
    [0.6905, 0.5385, 0.5789, 0.5185, 0.5, 0.5676]
]
top10_values = [
    [0.6905, 0.6538, 0.6842, 0.5185, 0.5, 0.6351],
    [0.7381, 0.6154, 0.6842, 0.5185, 0.5, 0.6351],
    [0.7619, 0.6538, 0.6842, 0.5556, 0.5294, 0.6486],
    [0.7857, 0.6923, 0.6842, 0.5556, 0.5, 0.6486]
]

colors=['#6CBAD8','#96C2D4','#BAD2E1','#D8E5F7']

# 绘图函数
def plot_bars(title, values):
    x = np.arange(len(datasets))
    bar_width = 0.13
    fig, ax = plt.subplots(figsize=(10, 6))
    
    for i, method_values in enumerate(values):
        ax.bar(x + i * bar_width, method_values, bar_width, label=methods[i], color=colors[i])
    
    ax.set_title(title, fontsize=16)
    ax.set_xlabel('Datasets', fontsize=14)
    ax.set_ylabel('Values', fontsize=14)
    ax.set_xticks(x + bar_width * (len(methods) - 1) / 2)
    ax.set_xticklabels(datasets, fontsize=12)
    ax.legend(title="Methods", fontsize=10, loc='upper right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

# 绘制柱状图
plot_bars("Top-1 Results", top1_values)
plot_bars("Top-5 Results", top5_values)
plot_bars("Top-10 Results", top10_values)
```
