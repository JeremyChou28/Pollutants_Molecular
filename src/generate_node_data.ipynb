```python
import numpy as np
import pandas as pd

data=pd.read_csv('../datasets/new_data/subset_5/edges.csv')

# Create two separate DataFrames: one for Library_node and one for Connected_node
library_nodes = data[['Library_node', 'SMILES']].rename(columns={'Library_node': 'ID'})
connected_nodes = data[['Connected_node', 'Expected_SMILES']].rename(columns={'Connected_node': 'ID', 'Expected_SMILES':'SMILES'})

# Concatenate the two DataFrames
merged_nodes = pd.concat([library_nodes, connected_nodes], ignore_index=True).drop_duplicates()

# Save the result to a new CSV file
output_path = '../datasets/new_data/subset_5/nodes_data.csv'
merged_nodes.to_csv(output_path, index=False)
print(len(merged_nodes))
```
