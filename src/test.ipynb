```python
import pandas as pd
from collections import Counter

nodes_data_1 = set(pd.read_csv(
    '../datasets/new_data/subset_1/nodes_data.csv')['ID'].values.tolist())
nodes_data_2 = set(pd.read_csv(
    '../datasets/new_data/subset_2/nodes_data.csv')['ID'].values.tolist())
nodes_data_3 = set(pd.read_csv(
    '../datasets/new_data/subset_3/nodes_data.csv')['ID'].values.tolist())
nodes_data_4 = set(pd.read_csv(
    '../datasets/new_data/subset_4/nodes_data.csv')['ID'].values.tolist())
nodes_data_5 = set(pd.read_csv(
    '../datasets/new_data/subset_5/nodes_data.csv')['ID'].values.tolist())

all_elements = nodes_data_1 | nodes_data_2 | nodes_data_3 | nodes_data_4 | nodes_data_5

# 统计每个元素在多少个集合中出现
element_counts = Counter()

for s in [nodes_data_1, nodes_data_2, nodes_data_3, nodes_data_4, nodes_data_5]:
    element_counts.update(s)

# 筛选出至少出现在2个集合中的元素
result = [element for element, count in element_counts.items() if count >= 2]

print(result)
```
